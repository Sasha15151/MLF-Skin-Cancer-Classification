#Importing Libraries
- os: Navigates file system
- python-dotenv: Reads the .env file so the paths stay private
- glob: Finds files using patterns
- pandas: Helps with working with tabular data
- numpy: Does computations
- matplotlib: Creates charts and plots. In this case, I'm using pyplot and image
- seaborn: Beautify charts
- PIL: Stands for Python Imaging Library. It opens images
- cv2: Stands for OpenCV. It processes images
- collections: Offers container datatypes of specialized, higher performance. In this case, I'm using Counter to count occurences of items (P.S. This is my first time looking into this library. I'm still considering whether to import it or drop it, so this list may change)
- warnings: Issues non-fatal messages about issues without stopping program execution, and allows programmers to modify how warnings can be handled
